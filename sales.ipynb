```python
#add values to null values in Region column based on the most frequent value for each CustomerID

df["Region"] = df["Region"].replace(r"^\s*$", pd.NA, regex=True)

df["Region"] = df.groupby("CustomerID")["Region"].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Unknown")
)
```

```python
df.isnull().sum()
```

```python
# adding values to null values in sales column based on product category

df["Sales"] = df["Sales"].fillna(
    df.groupby("ProductCategory")["Sales"].transform("median")
)

df["Sales"] = df["Sales"].fillna(df["Sales"].median())
```

```python
df.isnull().sum()
```

```python
df["Discount"] = df["Discount"].fillna(0.0)
```
